
# Ejemplo 4. Construcción a partir de factores

En este cuaderno estudiaremos la función de transferencia

\[
G(s)=\frac{10(1+s/10)}{s(1+s/2)}.
\]

La idea es construir su diagrama de Bode a partir de factores:

- una ganancia constante: \(K=10\),
- un integrador: \(1/s\),
- un polo real: \(1/(1+s/2)\),
- un cero real: \((1+s/10)\).

Nos concentraremos primero en la **magnitud**, mostrando:

1. la contribución de cada factor por separado,
2. sus aproximaciones asintóticas,
3. cómo estas contribuciones se van sumando,
4. y la comparación final entre el Bode exacto y el asintótico.


```python

import numpy as np
import matplotlib.pyplot as plt

# Frecuencia angular
w = np.logspace(-2, 3, 1200)

# Parámetros del ejemplo
K = 10
wp = 2      # polo real
wz = 10     # cero real
```


## 1. Factores individuales

Escribimos la función de transferencia como

\[
G(s)=K\left(\frac{1}{s}\right)\left(\frac{1}{1+s/2}\right)\left(1+\frac{s}{10}\right).
\]

En magnitud, al pasar a decibelios, las contribuciones se **suman**:

\[
M_G(\omega)=M_K(\omega)+M_{1/s}(\omega)+M_p(\omega)+M_z(\omega).
\]


```python

# ----- Magnitudes exactas (dB) -----
M_K = 20*np.log10(K) * np.ones_like(w)
M_int = -20*np.log10(w)
M_pole_exact = -10*np.log10(1 + (w/wp)**2)
M_zero_exact = 10*np.log10(1 + (w/wz)**2)

# ----- Aproximaciones asintóticas (dB) -----
M_K_asym = M_K.copy()
M_int_asym = M_int.copy()   # para el integrador, la magnitud exacta y la asintótica coinciden
M_pole_asym = np.where(w < wp, 0, -20*np.log10(w/wp))
M_zero_asym = np.where(w < wz, 0, 20*np.log10(w/wz))

# Totales
M_total_exact = M_K + M_int + M_pole_exact + M_zero_exact
M_total_asym = M_K_asym + M_int_asym + M_pole_asym + M_zero_asym
```

```python

plt.figure(figsize=(10, 6))
plt.semilogx(w, M_K_asym, label='Ganancia K=10', linewidth=2)
plt.semilogx(w, M_int_asym, label='Integrador 1/s', linewidth=2)
plt.semilogx(w, M_pole_asym, label='Polo 1/(1+s/2)', linewidth=2)
plt.semilogx(w, M_zero_asym, label='Cero 1+s/10', linewidth=2)
plt.axvline(wp, linestyle='--', linewidth=1)
plt.axvline(wz, linestyle='--', linewidth=1)
plt.xlabel(r'Frecuencia $\omega$ (rad/s)')
plt.ylabel('Magnitud (dB)')
plt.title('Aproximaciones asintóticas de los factores individuales')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```


## 2. Cómo se van sumando las contribuciones

Ahora construimos la magnitud asintótica paso a paso.

- La ganancia \(K=10\) aporta una recta horizontal de **20 dB**.
- El integrador \(1/s\) aporta una pendiente de **\(-20\) dB/década** desde el inicio.
- El polo en \(\omega_p=2\) rad/s reduce la pendiente en otras **\(-20\) dB/década**, de modo que la pendiente total pasa a ser **\(-40\) dB/década**.
- El cero en \(\omega_z=10\) rad/s aumenta la pendiente en **\(+20\) dB/década**, por lo que la pendiente final vuelve a ser **\(-20\) dB/década**.


```python

M_step1 = M_K_asym
M_step2 = M_K_asym + M_int_asym
M_step3 = M_K_asym + M_int_asym + M_pole_asym
M_step4 = M_K_asym + M_int_asym + M_pole_asym + M_zero_asym

plt.figure(figsize=(10, 6))
plt.semilogx(w, M_step1, label='Paso 1: K=10', linewidth=2)
plt.semilogx(w, M_step2, label='Paso 2: K + integrador', linewidth=2)
plt.semilogx(w, M_step3, label='Paso 3: + polo en 2 rad/s', linewidth=2)
plt.semilogx(w, M_step4, label='Paso 4: + cero en 10 rad/s', linewidth=3)
plt.axvline(wp, linestyle='--', linewidth=1)
plt.axvline(wz, linestyle='--', linewidth=1)
plt.xlabel(r'Frecuencia $\omega$ (rad/s)')
plt.ylabel('Magnitud (dB)')
plt.title('Construcción paso a paso de la magnitud asintótica')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```


## 3. Comparación entre magnitud exacta y magnitud asintótica

La magnitud exacta del sistema se obtiene a partir de

\[
G(j\omega)=\frac{10(1+j\omega/10)}{j\omega(1+j\omega/2)}.
\]

El diagrama exacto suaviza las transiciones cerca de las frecuencias de quiebre.


```python

plt.figure(figsize=(10, 6))
plt.semilogx(w, M_total_exact, label='Magnitud exacta', linewidth=3)
plt.semilogx(w, M_total_asym, '--', label='Magnitud asintótica', linewidth=2)
plt.axvline(wp, linestyle='--', linewidth=1)
plt.axvline(wz, linestyle='--', linewidth=1)
plt.xlabel(r'Frecuencia $\omega$ (rad/s)')
plt.ylabel('Magnitud (dB)')
plt.title('Magnitud exacta vs aproximación asintótica')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```


## 4. Verificación de las pendientes por regiones

Recordemos las pendientes esperadas:

- para \(\omega < 2\): pendiente total \(-20\) dB/década,
- para \(2 < \omega < 10\): pendiente total \(-40\) dB/década,
- para \(\omega > 10\): pendiente total \(-20\) dB/década.

La siguiente gráfica resalta estas regiones sobre la aproximación asintótica.


```python

plt.figure(figsize=(10, 6))
plt.semilogx(w, M_total_asym, linewidth=3, label='Magnitud asintótica total')
plt.axvline(wp, linestyle='--', linewidth=1, label=r'$\omega_p=2$ rad/s')
plt.axvline(wz, linestyle='--', linewidth=1, label=r'$\omega_z=10$ rad/s')

# Etiquetas de pendientes
plt.text(0.05, 42, '-20 dB/década')
plt.text(3.2, 6, '-40 dB/década')
plt.text(25, -34, '-20 dB/década')

plt.xlabel(r'Frecuencia $\omega$ (rad/s)')
plt.ylabel('Magnitud (dB)')
plt.title('Pendientes por regiones en la magnitud asintótica')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```


## 5. Fase exacta del sistema

La fase exacta está dada por

\[
\phi(\omega)=-90^\circ-\tan^{-1}(\omega/2)+\tan^{-1}(\omega/10).
\]

La mostramos a continuación para completar el ejemplo.


```python

phi_exact = -90 - np.degrees(np.arctan(w/wp)) + np.degrees(np.arctan(w/wz))

plt.figure(figsize=(10, 6))
plt.semilogx(w, phi_exact, linewidth=3)
plt.axvline(wp, linestyle='--', linewidth=1)
plt.axvline(wz, linestyle='--', linewidth=1)
plt.xlabel(r'Frecuencia $\omega$ (rad/s)')
plt.ylabel('Fase (grados)')
plt.title('Fase exacta del sistema')
plt.grid(True, which='both', alpha=0.3)
plt.show()
```


## 6. Comentarios finales

En este ejemplo observamos claramente la construcción del diagrama de Bode a partir de factores:

- la **ganancia** desplaza verticalmente la magnitud,
- el **integrador** fija la pendiente inicial en \(-20\) dB/década,
- el **polo** en \(2\) rad/s reduce aún más la pendiente,
- el **cero** en \(10\) rad/s compensa parcialmente ese efecto.

Este procedimiento es la base para construir de manera aproximada los diagramas de Bode de sistemas más complejos.
