# Cálculo de magnitudes y ángulos de los vectores

```python
import numpy as np

# Punto del plano complejo
s = -3 + 4j

# Función F(s)
F = (s + 1) / (s * (s + 2))

# Magnitud y ángulo
M = np.abs(F)
theta = np.angle(F, deg=True)

# Resultados
print("F(s) =", F)
print("Magnitud =", M)
print("Ángulo =", theta, "grados")
```
