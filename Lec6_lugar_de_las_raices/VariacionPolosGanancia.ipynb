# Variación de los polos con la ganancia

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def mostrar_polos(K):

    # Ecuación característica:
    # s^2 + 10s + K = 0
    polos = np.roots([1, 10, K])

    # Trayectoria recorrida desde K = 0 hasta el K seleccionado
    K_hist = np.linspace(0, max(K, 0.01), 300)

    polos_1 = []
    polos_2 = []

    for Ki in K_hist:
        p = np.roots([1, 10, Ki])
        polos_1.append(p[0])
        polos_2.append(p[1])

    polos_1 = np.array(polos_1)
    polos_2 = np.array(polos_2)

    # Figura
    plt.figure(figsize=(8, 6))

    # Trayectoria de los polos
    plt.plot(polos_1.real, polos_1.imag, '--', alpha=0.6)
    plt.plot(polos_2.real, polos_2.imag, '--', alpha=0.6)

    # Polos para el valor actual de K
    plt.scatter(
        polos.real,
        polos.imag,
        marker='x',
        s=120,
        linewidths=3,
        label='Polos de lazo cerrado'
    )

    # Ejes
    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)

    plt.xlim(-11, 1)
    plt.ylim(-11, 11)

    plt.xlabel('Parte real')
    plt.ylabel('Parte imaginaria')
    plt.title(f'Polos de lazo cerrado para K = {K:.1f}')
    plt.grid(True)
    plt.legend()

    plt.show()

    # Mostrar valores numéricos
    print("Polos de lazo cerrado:")

    for i, polo in enumerate(polos, start=1):
        print(f"Polo {i}: {polo.real:.3f} {polo.imag:+.3f}j")


# Control interactivo
slider_K = widgets.FloatSlider(
    value=0,
    min=0,
    max=100,
    step=1,
    description='K:',
    continuous_update=True
)

widgets.interactive(mostrar_polos, K=slider_K)
```
