## Problema 1

```python
import sympy as sp
import numpy as np
s = sp.symbols('s')
km, La, Ra, J, B, kv, K = sp.symbols('km La Ra J B kv K')
tf1 = 1/(s*La+Ra)
tf2 = 1/(J*s+B)
tf3 = tf1*km*tf2/(1+tf1*tf2/kv)
print(sp.simplify(tf3))
tf_seguimiento = 1/(1+tf3*K)
tf_seguimiento = sp.simplify(tf_seguimiento)
num, den = tf_seguimiento.as_numer_denom()
tf_seguimiento = sp.expand(num)/sp.expand(den)
print(tf_seguimiento)
```

## Problema 2

```python
tf_regulacion = tf2/(1+tf1*km*tf2*(K+1/kv))
tf_regulacion = sp.simplify(tf_regulacion)
num, den = tf_regulacion.as_numer_denom()
tf_regulacion = sp.expand(num)/sp.expand(den)
print(tf_regulacion)                 
```

## Problema 3

```python
tf4 = K*tf1*km*tf2/(1+tf1*km*tf2/kv)
tf_estabilidad = tf4/(1+tf4)
tf_estabilidad = sp.simplify(tf_estabilidad)
num, den = tf_estabilidad.as_numer_denom()
tf_estabilidad = sp.expand(num)/sp.expand(den)
print(tf_estabilidad)
tf_estabilidad = tf_estabilidad.subs([(La, 0.034/1000), (Ra, 0.108), (J, 77.6/(1000*10**2)), (km, 15.6/1000), (kv, 612*3.1416/30), (B, (274*10**-3)*(15.6*10**-3)/(7120*3.1416/30))])   
print(tf_estabilidad)
```

## Problema 4

```python
tf_seguimiento = tf_seguimiento.subs([(La, 0.034/1000), (Ra, 0.108), (J, 77.6/(1000*10**2)), (km, 15.6/1000), (kv, 612*3.1416/30), (B, (274*10**-3)*(15.6*10**-3)/(7120*3.1416/30))])   
Error_seguimiento = s*tf_seguimiento*6000*3.1416/30/s
Error_seguimiento = Error_seguimiento.subs([(s, 0)])
print(Error_seguimiento)
eq = Error_seguimiento - 0.1*6000*3.1416/30
K_sol = sp.solve(eq, K)
print(K_sol)
```

## Problema 5

```python
tf_regulacion = tf_regulacion.subs([(La, 0.034/1000), (Ra, 0.108), (J, 77.6/(1000*10**2)), (km, 15.6/1000), (kv, 612*3.1416/30), (B, (274*10**-3)*(15.6*10**-3)/(7120*3.1416/30)), (K, K_sol[0])])   
Td = sp.symbols('Td')
Error_regulacion = s*tf_regulacion*Td/s
Error_regulacion = Error_regulacion.subs([(s, 0)])
eq2 = Error_regulacion - 0.1*6000*3.1416/30
print(eq2)
Td_sol = sp.solve(eq2, Td)
print(Td_sol)
```
