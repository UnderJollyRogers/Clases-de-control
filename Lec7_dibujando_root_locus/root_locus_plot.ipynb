# Root locus plot

```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct


def findlines(artist):
    return (
        isinstance(artist, plt.Line2D)
        and artist.get_linestyle() == '-'
        and isinstance(artist.get_xdata(), np.ndarray)
    )


s = ct.TransferFunction.s

G = ((s + 3)*(s + 4)) / ((s + 1)*(s + 2)*(s**2 + s + 1))

print("G(s) =")
print(G)


plt.figure(figsize=(8, 6))

# El ; evita que Colab intente visualizar el objeto retornado
ct.rlocus(G, grid=True);

ax = plt.gca()

branches = ax.findobj(match=findlines)

for i, branch in enumerate(branches):
    branch.set_color(f'C{i % 10}')
    branch.set_linewidth(2)

plt.xlabel("Parte real")
plt.ylabel("Parte imaginaria")
plt.title("Root Locus")

plt.show()
```
